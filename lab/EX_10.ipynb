```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Step 1: Create Dataset
X = np.array([
    [1.0], [1.5], [2.0], [2.5], [3.0],
    [8.0], [8.5], [9.0], [9.5], [10.0]
])

# Step 2: Implement EM Algorithm using Gaussian Mixture Model
gmm = GaussianMixture(
    n_components=2,
    random_state=42
)

# Step 3: Train the Model
gmm.fit(X)

# Step 4: Predict Clusters
labels = gmm.predict(X)

# Step 5: Display Results
print("Expectation-Maximization Algorithm")
print("\nCluster Labels:")
print(labels)

print("\nCluster Means:")
print(gmm.means_)

print("\nCovariances:")
print(gmm.covariances_)

print("\nModel Converged:", gmm.converged_)

# Step 6: Plot the Clusters
plt.scatter(X, np.zeros(len(X)), c=labels, cmap='viridis',
            s=100, label='Data Points')

plt.scatter(gmm.means_, [0, 0], color='red',
            marker='X', s=200, label='Cluster Centers')

plt.title("Expectation-Maximization Algorithm")
plt.xlabel("Data Values")
plt.ylabel("Clusters")
plt.legend()
plt.grid()
plt.show()
```
